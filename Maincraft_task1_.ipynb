STEP 9

What is the average final grade (G3)?

```python
average_grade = df["G3"].mean()

print("Average final grade:", average_grade)
```

So the average final grade is approximately:

10.42

How many students scored above 15?

```python
students_above_15 = (df["G3"] > 15).sum()

print("Students who scored above 15:", students_above_15)
```

40 students scored above 15.

```python
correlation = df["studytime"].corr(df["G3"])

print("Correlation between study time and final grade:", correlation)
```

This is a very weak positive correlation and Students with higher study time tend to have slightly higher grades, but the relationship is very weak.

Which gender performs better on average?

```python
gender_average = df.groupby("sex")["G3"].mean()

print(gender_average)
```

Therefore : - Male students have a higher average final grade in this dataset.

**STEP 13 — Histogram of Grades**

A histogram helps us understand the distribution of students' final grades.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

### Study Time vs Final Grade

The scatterplot helps us visualize the relationship between study time and final grade.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="studytime",
    y="G3"
)

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

Because our correlation was only around 0.098, the points will not show a strong upward relationship.

Male vs Female Average Score



### Average Final Grade by Gender

This bar chart compares the average final grade of male and female students.

```python
gender_average = df.groupby("sex")["G3"].mean()

plt.figure(figsize=(7, 5))

gender_average.plot(kind="bar", edgecolor="black")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.xticks(rotation=0)

plt.show()
```

# **Final Conclusion**

After analyzing the Student Performance dataset, we found that the average final grade (G3) was approximately 10.42. There were 40 students who scored above 15.

The correlation between study time and final grade was approximately 0.098, which indicates a very weak positive relationship. Therefore, study time alone does not appear to have a strong relationship with final performance in this dataset.

When comparing gender, male students had a higher average final grade (approximately 10.91) than female students (approximately 9.97).
